```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read data
path = "../Raw/IMF_CPIS_WIDEF.csv"
df_cpis = pd.read_csv(path)
df_cpis
```

```python
indicators_to_keep = [
    "Assets, Total Investment",
    "Assets, Equity",
    "Assets, Debt Securities",
]

df_cpis = df_cpis[df_cpis["INDICATOR_LABEL"].isin(indicators_to_keep)]
```

```python
# Remove comp_3
df_cpis = df_cpis[df_cpis["COMP_BREAKDOWN_3_LABEL"] == "Sector: All sectors"].copy()
```

```python
df_cpis["INDICATOR_LABEL"].value_counts()
```

```python
# only numeric and 4 digits
year_cols = [col for col in df_cpis.columns if col.isdigit() and len(col) == 4]

df_cpis_long = df_cpis.melt(
    id_vars=["REF_AREA", "REF_AREA_LABEL", "INDICATOR", "INDICATOR_LABEL",
             "COMP_BREAKDOWN_1", "COMP_BREAKDOWN_1_LABEL",
             "COMP_BREAKDOWN_2", "COMP_BREAKDOWN_2_LABEL",
             "COMP_BREAKDOWN_3", "COMP_BREAKDOWN_3_LABEL",
             "UNIT_MEASURE", "UNIT_MEASURE_LABEL"],
    value_vars=year_cols,
    var_name="year",
    value_name="value"
)

df_cpis_long["year"] = df_cpis_long["year"].astype(int)
df_cpis_long = df_cpis_long.dropna(subset=["value"]).reset_index(drop=True)

df_cpis_long
```

```python
df_cpis_long["COMP_BREAKDOWN_1_LABEL"].value_counts()
```

```python
df_cpis_long["COMP_BREAKDOWN_2_LABEL"].value_counts()
```

```python
df_cpis_long["COMP_BREAKDOWN_3_LABEL"].value_counts()
```

```python
df_cpis_long["UNIT_MEASURE_LABEL"].value_counts()
```

```python
# Remove the "Counterpart:" in the COMP_BREAKDOWN_1_LABEL
df_cpis_long["COMP_BREAKDOWN_1_LABEL"] = df_cpis_long["COMP_BREAKDOWN_1_LABEL"].str.replace("Counterpart:", "", regex=False).str.strip()
# Remove the "Counterpart Sector:" in the COMP_BREAKDOWN_2_LABEL
df_cpis_long["COMP_BREAKDOWN_2_LABEL"] = df_cpis_long["COMP_BREAKDOWN_2_LABEL"].str.replace("Counterpart Sector:", "", regex=False).str.strip()
df_cpis_long.head()
```

```python
# Save to cleaned data
savepath = "../Clean/IMF_CPIS.csv"
df_cpis_long.to_csv(savepath, index=False)
```
