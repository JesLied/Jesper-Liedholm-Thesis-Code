```python
import requests
import pandas as pd
from io import StringIO

# ── Configuration ──────────────────────────────────────────────────────────────
BASE_URL = "https://sdmx.oecd.org/public/rest/data"
AGENCY    = "OECD.DAF.INV"
DATAFLOW  = "DSD_FDIRRI_SCORES@DF_FDIRRI_SCORES"
VERSION   = "2.0"
FILTER    = "._T..."          # all countries, total sector score
START     = "1997"            # data starts 1997

url = f"{BASE_URL}/{AGENCY},{DATAFLOW},{VERSION}/{FILTER}"

params = {
    "startPeriod": START,
    "format":      "csv",     # clean flat CSV, easiest to work with
}

headers = {
    "Accept":          "application/vnd.sdmx.data+csv; charset=utf-8; labels=both",
    "Accept-Encoding": "gzip, deflate, br",  # compression recommended by OECD docs
}

# ── Fetch ──────────────────────────────────────────────────────────────────────
response = requests.get(url, params=params, headers=headers, timeout=120)
response.raise_for_status()

df_raw = pd.read_csv(StringIO(response.text))

print("Raw columns:", df_raw.columns.tolist())
df_raw
```
