```python
import pandas as pd
```

```python
df = pd.read_csv("../Raw/dataset_2026-05-06T12_34_25.201892644Z_DEFAULT_INTEGRATION_IMF.MCM_FDI_1.0.0.csv")
df
```

```python
df["INDICATOR"].value_counts()
```

```python
# We want to use the Financial Development Index, since it is the aggregate.
df = df[df["INDICATOR"]=="Financial Development Index"]
```

```python
yr_cols = [c for c in df.columns if c.startswith("20") and len(c)==4]
yr_cols
```

```python
# Melt into long format
# Keep COUNTRY and INDICATOR as id_vars, and melt the year columns into 'year' and 'value'
df_long = df.melt(id_vars=["COUNTRY", "INDICATOR"], value_vars=yr_cols, var_name="year", value_name="value")
df_long["year"] = df_long["year"].astype(int)
df_long
```

```python
# Import pycountry and map them
import pycountry

name_to_iso3 = {country.name: country.alpha_3 for country in pycountry.countries}

df_long["iso3"] = df_long["COUNTRY"].map(name_to_iso3)
```

```python
# Expand with this one
EUROSTAT_COUNTRY_TO_ISO3 = {
    "Austria": "AUT",
    "Belgium": "BEL",
    "Bulgaria": "BGR",
    "Croatia": "HRV",
    "Cyprus": "CYP",
    "Czechia": "CZE",
    "Czech Republic": "CZE",
    "Denmark": "DNK",
    "Estonia": "EST",
    "Estonia, Republic of": "EST",
    "Finland": "FIN",
    "France": "FRA",
    "Germany": "DEU",
    "Greece": "GRC",
    "Hungary": "HUN",
    "Ireland": "IRL",
    "Italy": "ITA",
    "Latvia": "LVA",
    "Latvia, Republic of": "LVA",
    "Lithuania": "LTU",
    "Lithuania, Republic of": "LTU",
    "Luxembourg": "LUX",
    "Malta": "MLT",
    "Netherlands": "NLD",
    "Netherlands, The": "NLD",
    "Poland": "POL",
    "Poland, Republic of": "POL",
    "Portugal": "PRT",
    "Romania": "ROU",
    "Slovakia": "SVK",
    "Slovak Republic": "SVK",
    "Slovenia": "SVN",
    "Slovenia, Republic of": "SVN",
    "Spain": "ESP",
    "Sweden": "SWE",
    "United Kingdom": "GBR",
    "Norway": "NOR",
    "Switzerland": "CHE",
    "United States": "USA"
}

for country, iso3 in EUROSTAT_COUNTRY_TO_ISO3.items():
    df_long.loc[df_long["COUNTRY"] == country, "iso3"] = iso3
```

```python
# Get which ones dont map and print them
print("\n".join(df_long[df_long["iso3"].isna()]["COUNTRY"].unique()))
```

```python
# Rename value to financial_development_index
df_long = df_long.rename(columns={"value": "financial_development_index"})
df_long = df_long[["iso3", "year", "financial_development_index"]]
df_long
```
